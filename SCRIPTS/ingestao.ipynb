```python
import duckdb
import pandas as pd
import os
from datetime import datetime
```

```python
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
#utilizando os dados do arquivo z0019_1
arquivo='z0019_1.csv'
data_ingestao = datetime.now()
df= pd.read_csv(f'../landing/{arquivo}', sep=';')
df['nome_arquivo']= arquivo
df['data_ingestao'] = data_ingestao
df.head()
```

```python
con.execute("""
    CREATE TABLE IF NOT EXISTS BRONZE_PRODUTOS(
    NATBR VARCHAR,
    MAKYX VARCHAR, 
    WERKS VARCHAR, 
    MAINS VARCHAR,
    LABST VARCHAR,
    nome_arquivo VARCHAR, 
    data_ingestao TIMESTAMP
    )
""")
```

```python
con.execute("INSERT INTO BRONZE_PRODUTOS SELECT * FROM df")
```

```python
resultado= con.execute("SELECT * FROM BRONZE_PRODUTOS").fetch_df()
resultado.head()
```
